## 11. Data Cleaning

The dataset is cleaned before performing exploratory analysis.

The cleaning process includes:

- Removing unnecessary spaces from column names
- Removing duplicate records
- Cleaning text values
- Handling missing values
- Preparing numerical columns for analysis

```python
# Remove spaces from column names
df.columns = df.columns.str.strip()

# Remove duplicate rows
df = df.drop_duplicates()

print("Dataset shape after cleaning:", df.shape)
```

```python
# Remove extra spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text columns cleaned successfully.")
```

## 12. Column Renaming

Long column names are renamed to shorter and easier-to-use names.
This makes the analysis and visualization code easier to understand.

```python
df.rename(columns={
    "Estimated Unemployment Rate (%)": "Unemployment_Rate",
    "Estimated Employed": "Employed",
    "Estimated Labour Participation Rate (%)": "Labour_Participation"
}, inplace=True)

df.columns
```

## 13. Data Type Conversion

The Date column is converted into a datetime format so that
time-based analysis can be performed.

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

print(df["Date"].dtype)
```

```python
numeric_columns = [
    "Unemployment_Rate",
    "Employed",
    "Labour_Participation"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df[numeric_columns].dtypes
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=numeric_columns)

print("Dataset shape:", df.shape)
```

## 14. Feature Creation

Additional time-based features are created from the Date column.

These features will help us perform yearly and monthly analysis.

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()

df.head()
```

## 17. Time-Series Analysis

Time-series analysis is used to understand how unemployment
changed over the study period.

It helps identify:

- Increasing trends
- Decreasing trends
- Sudden changes
- Major fluctuations

```python
df = df.sort_values("Date")

df[["Date", "Unemployment_Rate"]].head()
```

```python
date_avg = (
    df.groupby("Date")["Unemployment_Rate"]
      .mean()
      .reset_index()
)

date_avg.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    date_avg["Date"],
    date_avg["Unemployment_Rate"],
    marker="o"
)

plt.title("Average Unemployment Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 22. COVID-19 Impact Analysis

The COVID-19 pandemic caused major disruptions to economic
activities and employment.

This section compares unemployment levels before and during
the COVID-19 period.

```python
df["COVID_Period"] = np.where(
    df["Date"] < pd.Timestamp("2020-03-01"),
    "Before COVID",
    "During COVID"
)

df[["Date", "COVID_Period"]].head()
```

```python
covid_avg = (
    df.groupby("COVID_Period")["Unemployment_Rate"]
      .mean()
)

covid_avg
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="COVID_Period",
    y="Unemployment_Rate"
)

plt.title("Unemployment Rate Before and During COVID-19")
plt.xlabel("COVID Period")
plt.ylabel("Average Unemployment Rate (%)")

plt.tight_layout()
plt.show()
```

```python
before = covid_avg["Before COVID"]
during = covid_avg["During COVID"]

change = during - before

percentage_change = ((during - before) / before) * 100

print(f"Before COVID: {before:.2f}%")
print(f"During COVID: {during:.2f}%")
print(f"Change: {change:.2f} percentage points")
print(f"Percentage change: {percentage_change:.2f}%")
```

## 23. Regional COVID-19 Impact

The change in unemployment between the pre-COVID and COVID
periods is analyzed for individual regions.

```python
regional_covid = (
    df.groupby(
        ["Region", "COVID_Period"]
    )["Unemployment_Rate"]
    .mean()
    .reset_index()
)

regional_covid.head()
```

```python
regional_covid_pivot = regional_covid.pivot(
    index="Region",
    columns="COVID_Period",
    values="Unemployment_Rate"
)

regional_covid_pivot.head()
```

```python
regional_covid_pivot["Change"] = (
    regional_covid_pivot["During COVID"]
    - regional_covid_pivot["Before COVID"]
)

regional_covid_pivot.sort_values(
    "Change",
    ascending=False
).head(10)
```

```python
top_covid_impact = (
    regional_covid_pivot["Change"]
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

top_covid_impact.sort_values().plot(kind="barh")

plt.title("Regions with Largest Increase in Unemployment During COVID-19")
plt.xlabel("Increase in Unemployment Rate (Percentage Points)")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

## 25. Outlier Analysis

Outliers represent observations that are unusually high or low
compared with the rest of the dataset.

An outlier is not automatically an error. It may represent a
genuine economic event or unusual observation.

```python
Q1 = df["Unemployment_Rate"].quantile(0.25)
Q3 = df["Unemployment_Rate"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Unemployment_Rate"] < lower_bound) |
    (df["Unemployment_Rate"] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```
