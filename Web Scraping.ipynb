**TASK 1: Web Scraping **

```python
import os
import pandas as pd
import time
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

all_data = []

for page in range(1, 51):

    url = f"https://www.ambitionbox.com/list-of-companies?page={page}"
    print("Scraping Page", page)

    driver.get(url)
    time.sleep(6)

    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)

    soup = BeautifulSoup(driver.page_source, "html.parser")

    cards = soup.select("div.companyCardWrapper")

    for card in cards:
        try:
            # Company Name
            company = card.select_one("h2")
            company = company.text.strip() if company else ""

            # ✅ FIXED RATING (NEW SELECTOR)
            rating = card.select_one("div.rating, span[class*='rating']")
            rating = rating.text.strip() if rating else ""

            # Reviews / Salary / Interviews / Jobs
            stats = card.select("span.companyCardWrapper__ActionCount")

            reviews = stats[0].text.strip() if len(stats) > 0 else ""
            salary = stats[1].text.strip() if len(stats) > 1 else ""
            interviews = stats[2].text.strip() if len(stats) > 2 else ""
            jobs = stats[3].text.strip() if len(stats) > 3 else ""

            # Industry + Location
            info = card.select_one("span.companyCardWrapper__interLinking")

            industry = ""
            location = ""

            if info:
                text = info.text.strip().split("|")
                industry = text[0].strip() if len(text) > 0 else ""
                location = text[1].strip() if len(text) > 1 else ""

            # ✅ FIXED BENEFITS (NEW APPROACH)
            benefits = ""
            benefit_block = card.find_all("span")

            for b in benefit_block:
                if "benefit" in b.text.lower():
                    benefits = b.text.strip()
                    break

            all_data.append({
                "Company": company,
                "Rating": rating,
                "Reviews": reviews,
                "Salary": salary,
                "Interviews": interviews,
                "Jobs": jobs,
                "Benefits": benefits,
                "Industry": industry,
                "Location": location
            })

        except:
            continue

driver.quit()

df = pd.DataFrame(all_data)

save_path = os.path.join(os.path.expanduser("~"), "Downloads", "ambitionbox_companies_fixed.csv")
df.to_csv(save_path, index=False)

print("Total Companies:", len(df))
print("Saved at:", save_path)
print(df.head())
```
